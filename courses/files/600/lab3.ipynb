Facts you will use. Nakamura's rating 3300, opponents 2934, so $d=366$. A 400-point Elo gap is 10-to-1, so the win probability for one game is

$$
q=\frac{10^{d/400}}{1+10^{d/400}}.
$$

Count the draw as no-win. Then $E$ is 45 wins in 46 independent games. That is binomial:

$$
p=P(E\mid I)=\binom{46}{45}q^{45}(1-q).
$$
