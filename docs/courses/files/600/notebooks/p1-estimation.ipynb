## A. The same estimate, different information (10 minutes)
Predict: will 8/10 and 80/100 have the same likelihood maximizer?
Which curve will be narrower? Explain before running the cell.

```python
p = np.linspace(.001, .999, 999)
fig, ax = plt.subplots()
for successes, n in [(8, 10), (80, 100)]:
    log_likelihood = successes*np.log(p) + (n-successes)*np.log1p(-p)
    relative_likelihood = np.exp(log_likelihood-log_likelihood.max())
    ax.plot(p, relative_likelihood, label=f"{successes}/{n}")
ax.set(xlabel="Unknown probability p", ylabel="Relative likelihood")
ax.legend()
plt.show()
```

## B. Repeating the experiment (35 minutes, including Monte Carlo check)
Predict the standard error at n=100 relative to n=25.
Then change sample size while keeping the generating probability fixed.

```python
rng = np.random.default_rng(600)
true_p = .6
sample_sizes = [25, 100, 400]
fig, axes = plt.subplots(1, 3, figsize=(12, 3.5), sharex=True)
for ax, n in zip(axes, sample_sizes):
    estimates = rng.binomial(n, true_p, size=5000)/n
    empirical_se = estimates.std(ddof=1)
    theoretical_se = np.sqrt(true_p*(1-true_p)/n)
    ax.hist(estimates, bins=np.linspace(0, 1, 31), density=True)
    ax.axvline(true_p, color="black")
    ax.set(title=f"n={n}, SD={empirical_se:.3f}", xlabel="Estimated p")
    print(n, "simulation SD", round(empirical_se, 4),
          "theory", round(theoretical_se, 4))
plt.tight_layout()
plt.show()
```

### Last five minutes of B: an expectation by simulation
Y takes values 0,1,3 with probabilities .2,.5,.3. Let g(Y)=(Y-1)^2.
Calculate E[g(Y)] by hand, then predict whether g(E[Y]) is the same.

```python
values = np.array([0., 1., 3.])
weights = np.array([.2, .5, .3])
exact = weights @ ((values-1)**2)
draws = np.random.default_rng(600).choice(values, size=20000, p=weights)
estimate = np.mean((draws-1)**2)
print("Exact E[g(Y)]:", exact, "Monte Carlo:", estimate)
print("g(E[Y]):", (weights @ values-1)**2)
assert np.isclose(exact, 1.4)
```

The law of large numbers explains convergence of an average under assumptions.
