## 1. Rayleigh's argon: a difference too big to be noise

- Lord Rayleigh weighed the same volume of nitrogen from two sources.
- One sample came from air, after he removed the oxygen.
- The other came from a chemical reaction, decomposing nitrous oxide and nitric oxide.
- The numbers are densities, in grams per liter.
- One air measurement is missing, so air has 7 values and the chemical source has 8.

- The two-sample $t$-statistic measures that gap in units of its standard error.
- The two groups have very different spreads, printed as standard deviations.
- So we use Welch's version, which does not assume one common spread:

$$ t = \frac{\bar x_1 - \bar x_2}{\sqrt{s_1^2/n_1 + s_2^2/n_2}}. $$

- Here $\bar x$ is a mean, $s$ a standard deviation, and $n$ a sample size.
- A large $|t|$, say above 2, means the gap is large next to pure noise.


```python
air = np.array([2.31017, 2.30986, 2.31010, 2.31001, 2.31024, 2.31010, 2.31028])
decomp = np.array([2.30143, 2.29890, 2.29816, 2.30182, 2.29869, 2.29940, 2.29849, 2.29889])
print(f"air   : n={len(air)}, mean={air.mean():.5f}, sd={air.std(ddof=1):.6f}")
print(f"decomp: n={len(decomp)}, mean={decomp.mean():.5f}, sd={decomp.std(ddof=1):.6f}")

n1, n2 = len(air), len(decomp)
se_welch = np.sqrt(air.var(ddof=1) / n1 + decomp.var(ddof=1) / n2)
t_by_hand = (air.mean() - decomp.mean()) / se_welch
res = stats.ttest_ind(air, decomp, equal_var=False)            # Welch
res_pooled = stats.ttest_ind(air, decomp, equal_var=True)      # assumes one common spread
print(f"Welch t by hand = {t_by_hand:.1f}, scipy Welch t = {res.statistic:.1f}, p-value = {res.pvalue:.1e}")
print(f"(the pooled version, which assumes a common spread, gives t = {res_pooled.statistic:.1f})")
```

**Reading.**

- The means are 2.31 and 2.30. They look close.
- The spread inside each group is tiny: standard deviations about 0.0001 and 0.0014.
- Welch's $t$ is about 21.7, and the p-value is about 8e-08.
- The pooled version gives 20.2, also far above 2.
- Rayleigh took the gap seriously. The air sample held a little of a heavier gas: argon.


```python
# Your turn: keep only a fraction of the real gap and watch t fall.
# Target: find a value of frac where t drops below 2.
frac = 1.0   # change this: 1.0 = the real data, 0.0 = no gap at all
shifted = decomp - (1 - frac) * (decomp.mean() - air.mean())   # slides decomp toward air
t_new = stats.ttest_ind(air, shifted, equal_var=False).statistic
print(f"frac = {frac}: gap = {air.mean() - shifted.mean():.5f}, t = {t_new:.1f}")
```

**Check yourself.**

1. The two means differ by only about 0.5%. Why is the $t$-statistic so large?
2. If you doubled the measurements in each group, and the gap and spread stayed the same, would $t$ go up or down?

1. The standard error is tiny, because the measurements inside a group are very close. $t$ compares the gap with the noise, not with the size of the numbers.
2. Up, by about $\sqrt 2$. The standard error shrinks like $1/\sqrt n$.

</details>


## 2. A/B testing: SimCity

- Electronic Arts tested the pre-order page for SimCity.
- One page kept a promotional banner. The other took the banner off.

![Original, with the offer](simcity_original.jpg)

![Variation, the offer removed](simcity_variation.jpg)

- The page with no offer drove more purchases.
- The numbers below are the book's hypothetical follow-up: one week, a pink banner against a black one.
- Pink banner: 40% of 500 visitors bought.
- Black banner: 30% of 550 visitors bought.

- Let $p_1$ be the true purchase rate on pink, and $p_2$ the true rate on black.
- We see only the sample rates $\hat p_1 = 0.40$ and $\hat p_2 = 0.30$.

- One rate: if $n$ visitors each buy with probability $p$, the standard error of $\hat p$ is $\sqrt{p(1-p)/n}$.
- We plug in $\hat p$ for the unknown $p$.

- The two groups are different visitors, so the two errors are independent and the variances add:

$$ \mathrm{SE}_{\text{diff}} = \sqrt{\frac{\hat p_1(1-\hat p_1)}{n_1} + \frac{\hat p_2(1-\hat p_2)}{n_2}}, \qquad \text{95\% interval: } (\hat p_1 - \hat p_2) \pm 1.96\, \mathrm{SE}_{\text{diff}}. $$

- The null says both pages have the same rate.
- Pool all visitors to estimate that one rate: $\hat p_0 =$ (all buys) / (all visitors).

$$ z = \frac{\hat p_1 - \hat p_2}{\sqrt{\hat p_0 (1-\hat p_0)(1/n_1 + 1/n_2)}}. $$

- A $|z|$ above 1.96 rejects the null at 5%.
- The two-sided p-value is $2\,(1 - \Phi(|z|))$, where $\Phi$ is the standard normal distribution function.

- Open the [A/B Test Lab](https://vsokolov.org/labs/ab-test-lab.html) and run 40% against 30%, with 500 and 550 visitors.
- Then open the [Confidence Interval Lab](https://vsokolov.org/labs/ci-coverage-lab.html) and repeat the experiment.
- About 95 of 100 intervals should contain the true value. That is what "95%" means.

- Do not decide by whether two separate intervals overlap.
- The standard error of the difference is $\sqrt{\mathrm{SE}_1^2 + \mathrm{SE}_2^2}$, which is smaller than $\mathrm{SE}_1 + \mathrm{SE}_2$.


```python
# SimCity: one week of visitors (from the book: pink 40% of 500, black 30% of 550)
n_pink, buys_pink = 500, 200
n_black, buys_black = 550, 165
rate_pink = buys_pink / n_pink
rate_black = buys_black / n_black
print(f"pink : {buys_pink} buys of {n_pink} visitors, rate {rate_pink:.3f}")
print(f"black: {buys_black} buys of {n_black} visitors, rate {rate_black:.3f}")
```

```python
zc = stats.norm.ppf(0.975)                      # 1.96
se1 = np.sqrt(rate_pink * (1 - rate_pink) / n_pink)
se2 = np.sqrt(rate_black * (1 - rate_black) / n_black)
se_diff = np.sqrt(se1**2 + se2**2)
diff = rate_pink - rate_black
print(f"difference = {diff:.3f}, SE of difference = {se_diff:.4f}")
print(f"95% interval for the difference: ({diff - zc * se_diff:.3f}, {diff + zc * se_diff:.3f})")

p0 = (buys_pink + buys_black) / (n_pink + n_black)
z_stat = diff / np.sqrt(p0 * (1 - p0) * (1 / n_pink + 1 / n_black))
print(f"pooled rate = {p0:.3f}, z = {z_stat:.2f}, p-value = {2 * (1 - stats.norm.cdf(abs(z_stat))):.1e}")
print(f"separate 95% intervals: pink ({rate_pink - zc * se1:.3f}, {rate_pink + zc * se1:.3f}), black ({rate_black - zc * se2:.3f}, {rate_black + zc * se2:.3f})")
print(f"SE1 + SE2 = {se1 + se2:.4f}, but SE of the difference = {se_diff:.4f}")
```

**Reading.**

- The 95% interval for the difference is (0.042, 0.158), and it stays above zero.
- The $z$-statistic is 3.40, so the p-value is about 0.0007.
- The last line of the output: $\mathrm{SE}_1 + \mathrm{SE}_2$ is larger than the standard error of the difference.

- A lift is the rise in the purchase rate, $p_{\text{new}} - p_{\text{old}}$. From 30% to 32% is a lift of 0.02.
- How many visitors do we need before the test, so a small real lift would be found?

- Let $p_{\text{old}}$ be the current purchase rate, and $p_{\text{new}}$ the rate after the lift.
- The formula detects that lift at 5% significance, two-sided, and 80% power.
- Power here means the test finds the lift 80% of the time when it is real.

$$ n = \frac{(z_{0.975} + z_{0.80})^2 \,\big[p_{\text{old}}(1-p_{\text{old}}) + p_{\text{new}}(1-p_{\text{new}})\big]}{(p_{\text{new}} - p_{\text{old}})^2}. $$

```python
# Your turn: the formula as a function. The cell runs with a 5-point lift from 30%.
# Target: set `lift = 0.025` (half as big) and compare n with the value for 0.05.
def n_per_arm(p_start, lift, power=0.80):
    p_end = p_start + lift
    z_sum = stats.norm.ppf(0.975) + stats.norm.ppf(power)
    return int(np.ceil(z_sum**2 * (p_start * (1 - p_start) + p_end * (1 - p_end)) / lift**2))

lift = 0.05   # change this
print(f"lift {lift:.3f} from 30%: n = {n_per_arm(0.30, lift)} visitors per arm")
```

## 3. Peeking and multiple testing

The first split is the truth. The second is what the test says. The leaf is the name of that pair.

![Tree of the two errors. Truth splits into yes and no. Each of those splits into predicted yes and predicted no. The leaves are TP, Type II, Type I, and TN.](error-tree.png)

- Type I is a false alarm: the truth is no, and the test says yes.
- Type II is a missed discovery: the truth is yes, and the test says no.

- A p-value is a number from this one sample. Call a discovery only if $p < 0.05$, fixed before you look.

![When there is no real effect, the two tails beyond 1.96 are 5% together.](type1-cutoff.png)

- Bonferroni divides 0.05 by the number of tests.

![A larger sample moves the real effect past the line, so less of it is missed.](type2-sample.png)

- With the cutoff fixed, the sample size is what sets Type II. Section 2 picks it so the effect is found 80% of the time.
- A stricter cutoff misses more real effects, unless the sample grows too.

![The Economist, 2013, at 40% power. Of 85 discoveries, 45 are false alarms.](economist-counts.png)

- Peeking: test every 100 visitors per arm, and stop at the first $|z| > 1.96$.
- Many variants: 20 banners, and ship any one with $|z| > 1.96$. [Multiple Testing Lab](https://vsokolov.org/labs/multiple-testing-lab.html).

![One fair experiment. Stopping at the first look keeps a false alarm. Waiting until the end, the statistic is back inside.](peeking-path.png)

- Twenty looks means twenty chances. The chance of at least one crossing is far above 5%. The next cell counts how often.
- The check is an A/A test: both arms truly buy at rate 0.30, so every winner is a false alarm.

### Computing corner: vectorized simulation

We draw all 1,000 experiments in one array of shape (1000 experiments, 20 looks); cumulative sums along the looks give the running totals.


```python
import time

def peeking_vectorized(n_exp=1000, looks=20, block=100, rate=0.30, seed=11):
    rng = np.random.default_rng(seed)
    buys1 = rng.binomial(block, rate, size=(n_exp, looks)).cumsum(axis=1)   # arm A running buys
    buys2 = rng.binomial(block, rate, size=(n_exp, looks)).cumsum(axis=1)   # arm B running buys
    n_seen = block * np.arange(1, looks + 1)                                # visitors per arm so far
    p_hat = (buys1 + buys2) / (2 * n_seen)
    z = (buys1 - buys2) / n_seen / np.sqrt(p_hat * (1 - p_hat) * 2 / n_seen)
    return z

def peeking_loop(n_exp=1000, looks=20, block=100, rate=0.30, seed=11):
    """The beginner's version: two loops, one random draw at a time. Returns the share of false positives."""
    rng = np.random.default_rng(seed)
    n_false = 0
    for e in range(n_exp):
        b1 = b2 = 0
        for look in range(1, looks + 1):
            b1 += rng.binomial(block, rate)
            b2 += rng.binomial(block, rate)
            n_seen = block * look
            p_hat = (b1 + b2) / (2 * n_seen)
            z = (b1 - b2) / n_seen / np.sqrt(p_hat * (1 - p_hat) * 2 / n_seen)
            if abs(z) > 1.96:
                n_false += 1
                break                                    # stop at the first look that crosses
    return n_false / n_exp

t0 = time.perf_counter(); z_vec = peeking_vectorized(); t_vec = time.perf_counter() - t0
t0 = time.perf_counter(); fp_loop = peeking_loop(); t_loop = time.perf_counter() - t0
print(f"vectorized: {t_vec * 1000:.1f} ms, double loop: {t_loop * 1000:.1f} ms, the loop is {t_loop / t_vec:.0f} times slower")

fp_end = (np.abs(z_vec[:, -1]) > 1.96).mean()            # test once, at 2,000 visitors per arm
fp_peek = (np.abs(z_vec) > 1.96).any(axis=1).mean()       # stop at the first look that crosses
print(f"false positives, testing once at the end : {fp_end:.3f}")
print(f"false positives, peeking after every 100 : {fp_peek:.3f}")
print(f"false positives, peeking, double-loop version : {fp_loop:.3f}")
```

**Reading.**

- Testing once at the end gives a false-positive rate of 0.051, close to the promised 0.05.
- Peeking after every 100 visitors per arm, 20 looks, gives 0.234.
- The double loop, one draw at a time, is several times slower than the array version.
- Write simulations as array operations. Loop only over a step that depends on the one before it, such as a bandit's visitors.

### Many variants

- Now 20 variants against one control, 1,000 visitors per arm.
- Every arm has the same true rate, 0.30.
- Count how many variants look significant in each experiment.


```python
def many_variants(n_exp=1000, k=20, n=1000, rate=0.30, seed=12):
    rng = np.random.default_rng(seed)
    control = rng.binomial(n, rate, size=(n_exp, 1))
    variants = rng.binomial(n, rate, size=(n_exp, k))
    p_hat = (control + variants) / (2 * n)
    z = (variants - control) / n / np.sqrt(p_hat * (1 - p_hat) * 2 / n)
    return 2 * (1 - stats.norm.cdf(np.abs(z)))          # p-values, shape (n_exp, k)

pv = many_variants()
sig = (pv < 0.05).sum(axis=1)
print(f"average number of false winners per experiment: {sig.mean():.2f}")
print(f"share of experiments with at least one false winner: {(sig >= 1).mean():.3f}  (theory: {1 - 0.95**20:.3f})")
print(f"Bonferroni (cutoff 0.05/20 = {0.05 / 20}): share with at least one false winner: {(pv < 0.05 / 20).any(axis=1).mean():.3f}")
```

**Reading.**

- With 20 variants, an experiment has on average 0.94 false winners.
- And 0.413 of experiments show at least one.
- For 20 independent tests the theory says $1 - 0.95^{20} = 0.642$.
- The simulation is lower because the variants share one control, so false winners tend to come together.
- Bonferroni divides the cutoff by the number of tests: $0.05/20 = 0.0025$.
- That brings the chance of any false winner down to about 0.025. It guarantees at most 0.05.
- A true difference then needs a larger $z$ to be found.

## 4. Bayes learning: the same test with Beta posteriors

- On a grid of possible values, the posterior is the prior times the likelihood. Divide by the sum so the bars add to 1:

$$ p(\theta \mid y) = \frac{p(y \mid \theta)\, p(\theta)}{\sum_\theta p(y \mid \theta)\, p(\theta)}. $$

- Ten coin flips, three of them tails. $\theta$ is the probability of tails, and the likelihood is $\theta^3(1-\theta)^7$.
- The prior is largest at a fair coin, $\theta = 0.5$, and it is zero at 0 and at 1.
- The last panel keeps this prior and uses 100 flips, 30 of them tails.


```python
theta = np.round(np.arange(0, 1.01, 0.1), 1)
prior = np.array([0, 0.024, 0.077, 0.132, 0.173, 0.188, 0.173, 0.132, 0.077, 0.024, 0])

def grid_likelihood(n, y):
    out = np.zeros_like(theta, dtype=float)
    ok = (theta > 0) & (theta < 1)
    out[ok] = np.exp(y * np.log(theta[ok]) + (n - y) * np.log(1 - theta[ok]))
    return out

def grid_posterior(n, y):
    raw = grid_likelihood(n, y) * prior
    return raw / raw.sum()

lik10 = grid_likelihood(10, 3)
lik10 = lik10 / lik10.sum()
post10 = grid_posterior(10, 3)
post100 = grid_posterior(100, 30)

fig, axes = plt.subplots(1, 4, figsize=(10, 2.4))
for ax, height, title in zip(axes, [prior, lik10, post10, post100],
        ["prior", "likelihood, sums to 1", "posterior, 10 flips", "posterior, 100 flips"]):
    ax.bar(theta, height, width=0.08, color="black")
    ax.set_title(title, fontsize=9)
    ax.set_xlabel(r"$\theta$")
    ax.set_xticks(theta[::2])
axes[0].set_ylabel("probability")
fig.tight_layout()
plt.show()
print("posterior mean, 10 flips:", round((theta * post10).sum(), 3), "   count 3/10 = 0.3")
print("posterior mean, 100 flips:", round((theta * post100).sum(), 3), "   count 30/100 = 0.3")
```

**Reading.**

- After 10 flips the posterior mean is 0.375, between the prior mean 0.5 and the count 0.30.
- After 100 flips the mean is 0.31, and the bars sit on 0.3.


- The z-test gave an estimate, an interval, a statistic, and a p-value.
- A Bayesian analysis gives a whole distribution for each purchase rate.

- A Beta($a$, $b$) distribution lives on $[0, 1]$, so it fits a rate $p$.
- Its mean is $a/(a+b)$.
- Beta(1, 1) is flat: every rate is equally likely before the data. This is the prior.

- After $n$ visitors, with $s$ buys and $n - s$ non-buys, the posterior is our belief after the data:

$$ \text{Beta}(a, b) \;\longrightarrow\; \text{Beta}(a + s,\; b + n - s). $$

- From Beta(1, 1), pink (200 buys of 500) gives Beta(201, 301).
- Black (165 buys of 550) gives Beta(166, 386).
- Draw many values from each posterior and count how often the pink draw is larger.


```python
a0, b0 = 1, 1                                   # the flat prior, Beta(1, 1)
post_pink = (a0 + buys_pink, b0 + n_pink - buys_pink)
post_black = (a0 + buys_black, b0 + n_black - buys_black)
print("posterior pink : Beta", post_pink, " mean", round(post_pink[0] / sum(post_pink), 3))
print("posterior black: Beta", post_black, " mean", round(post_black[0] / sum(post_black), 3))

rng_b = np.random.default_rng(3)
draw_pink = rng_b.beta(*post_pink, size=100_000)
draw_black = rng_b.beta(*post_black, size=100_000)
print("P(pink > black) by sampling:", (draw_pink > draw_black).mean())
print("95% credible interval for pink :", np.percentile(draw_pink, [2.5, 97.5]).round(3))
print("95% credible interval for black:", np.percentile(draw_black, [2.5, 97.5]).round(3))

grid = np.linspace(0, 1, 500)
plt.plot(grid, stats.beta.pdf(grid, a0, b0), color="gray", linestyle=":", label="prior Beta(1, 1)")
plt.plot(grid, stats.beta.pdf(grid, *post_pink), color="C3", label="pink posterior")
plt.plot(grid, stats.beta.pdf(grid, *post_black), color="C0", linestyle="--", label="black posterior")
plt.xlim(0.15, 0.55)
plt.xlabel("purchase rate")
plt.ylabel("density")
plt.legend()
plt.show()
```

## 6. Bandits: Thompson sampling, the A/B test that learns while it runs

- An A/B test learns, then acts. It splits traffic 50/50 for the whole test, and picks a winner afterward.
- A bandit learns while it acts. It sends more visitors to the page that looks better, and keeps learning from the rest.
- This is explore and exploit: try other options to learn, or use the best one so far to earn.

- Thompson sampling keeps a Beta posterior for each page, starting from Beta(1, 1). For each new visitor, draw one value from each posterior and show the page with the larger draw:

$$ \tilde\theta_k \sim \mathrm{Beta}(\alpha_k, \beta_k), \qquad a = \arg\max_k \tilde\theta_k. $$

- The visitor buys or not, $y \in \{0, 1\}$. Add that count to the page that was shown:

$$ \alpha_a \leftarrow \alpha_a + y, \qquad \beta_a \leftarrow \beta_a + (1 - y). $$

- A page that is still uncertain has a wide posterior, so it sometimes wins the draw and gets tried again.

- Open the [Bandit vs A/B Lab](https://vsokolov.org/labs/bandit-lab.html) and run Thompson sampling against the 50/50 split.
- Use rates 0.40 and 0.30, and watch the traffic shift to the better page.

- We compare the two in simulation. The true rates are pink 0.40 and black 0.30. The algorithm does not know them.
- We run 200 independent experiments of 5,000 visitors at once.
- The loop is over visitors. Each step is an array operation over the 200 experiments.
- Regret is the purchases lost against an oracle that always shows pink. The true rates are $r_0 = 0.40$ and $r_1 = 0.30$, and $a_t$ is the page shown to visitor $t$:

$$ \mathrm{regret} = \sum_{t=1}^{T} (0.40 - r_{a_t}). $$


**Predict.**

- Does the Thompson bandit ever stop sending visitors to the worse page?
- After 5,000 visitors, what share of traffic do you expect on pink: 60%, 90%, or 100%?


```python
def run_policy(policy, T=5000, R=200, seed=0, rates=(0.40, 0.30), eps=0.1, alpha=None):
    """Return the arm shown at each visitor, shape (R, T). Arm 0 = pink, arm 1 = black."""
    rng = np.random.default_rng(seed)
    rates = np.array(rates)
    rows = np.arange(R)
    a = np.ones((R, 2)); b = np.ones((R, 2))          # Beta counts (Thompson)
    Q = np.zeros((R, 2)); cnt = np.zeros((R, 2))      # running estimates (epsilon-greedy)
    arms = np.zeros((R, T), dtype=int)
    for t in range(T):
        if policy == "ab":
            arm = rng.integers(0, 2, R)
        elif policy == "thompson":
            arm = rng.beta(a, b).argmax(axis=1)
        else:  # epsilon-greedy
            greedy = (Q + 1e-9 * rng.random((R, 2))).argmax(axis=1)
            arm = np.where(rng.random(R) < eps, rng.integers(0, 2, R), greedy)
        r = (rng.random(R) < rates[arm]).astype(float)
        a[rows, arm] += r
        b[rows, arm] += 1 - r
        cnt[rows, arm] += 1
        step = (1 / cnt[rows, arm]) if alpha is None else alpha
        Q[rows, arm] += step * (r - Q[rows, arm])     # Q <- Q + alpha (r - Q)
        arms[:, t] = arm
    return arms

true_rates = np.array([0.40, 0.30])
arms_ab = run_policy("ab", seed=60)
arms_ts = run_policy("thompson", seed=61)

def regret_curve(arms):
    return (0.40 - true_rates[arms]).cumsum(axis=1).mean(axis=0)       # average over the 200 experiments

def pink_share(arms, window=100):
    x = (arms == 0).mean(axis=0)
    return np.convolve(x, np.ones(window) / window, mode="valid")

fig, ax = plt.subplots(1, 2, figsize=(9, 3.5))
ax[0].plot(pink_share(arms_ab), color="C0", linestyle="--", label="50/50 test")
ax[0].plot(pink_share(arms_ts), color="C3", label="Thompson sampling")
ax[0].set_xlabel("visitor number"); ax[0].set_ylabel("share of visitors shown pink"); ax[0].legend()
ax[1].plot(regret_curve(arms_ab), color="C0", linestyle="--", label="50/50 test")
ax[1].plot(regret_curve(arms_ts), color="C3", label="Thompson sampling")
ax[1].set_xlabel("visitor number"); ax[1].set_ylabel("purchases lost (regret)"); ax[1].legend()
plt.tight_layout(); plt.show()

print(f"regret after 5,000 visitors: 50/50 test {regret_curve(arms_ab)[-1]:.0f}, Thompson {regret_curve(arms_ts)[-1]:.0f}")
print(f"share shown pink, visitors 1 to 500   : 50/50 {(arms_ab[:, :500] == 0).mean():.3f}, Thompson {(arms_ts[:, :500] == 0).mean():.3f}")
print(f"share shown pink, last 500 visitors   : 50/50 {(arms_ab[:, -500:] == 0).mean():.3f}, Thompson {(arms_ts[:, -500:] == 0).mean():.3f}")
```

**Reading.**

- The 50/50 test keeps sending half the visitors to each page, and its regret grows in a straight line.
- After 5,000 visitors it has lost 250 purchases: half the visitors see black, and each loses 0.1 on average.
- Thompson sampling starts at 50/50, because both posteriors are wide, then it shifts.
- In the last 500 visitors, pink gets 99.3% of the traffic.
- Its regret is 16, and the curve flattens.
- The chance of showing black stays above zero at every visitor, because black's posterior is never certain.
- That chance shrinks as the evidence grows.
- The curves are averages over 200 runs. One run can go a long stretch without showing black.

- Epsilon-greedy keeps a running estimate $Q$ of each page's rate.
- After a visitor with reward $r$, 1 for a buy and 0 for no buy:

$$ Q \leftarrow Q + \alpha\,(r - Q). $$

- $\alpha$ is the step size.
- With probability $1-\varepsilon$ it shows the page with the larger $Q$. That is exploit.
- With probability $\varepsilon$ it shows a random page. That is explore. Here $\varepsilon = 0.1$.

- In Q-learning the target also includes the reward expected after the action. Here the target is just the observed reward, because this is a bandit.
- It is also stochastic gradient descent on the squared error $\tfrac12 (r - Q)^2$.
- The gradient in $Q$ is $-(r-Q)$, so a step of size $\alpha$ is the update above.
- With $\alpha = 1/n$, where $n$ is how many visitors have seen that page, $Q$ is exactly the running average.

- A hand trace. One page, $Q$ starts at 0, rewards 1, 0, 1, and $\alpha = 1/n$.

| visitor $n$ | reward $r$ | $\alpha = 1/n$ | new $Q = Q + \alpha (r - Q)$ |
|---|---|---|---|
| 1 | 1 | 1 | $0 + 1 \times (1 - 0) = 1$ |
| 2 | 0 | 1/2 | $1 + \tfrac12 (0 - 1) = 0.5$ |
| 3 | 1 | 1/3 | $0.5 + \tfrac13 (1 - 0.5) = 0.6667$ |

- With a constant step, such as $\alpha = 0.2$, old rewards fade instead of being averaged equally.


```python
arms_eg = run_policy("eps", seed=62)
print(f"regret after 5,000 visitors: 50/50 {regret_curve(arms_ab)[-1]:.0f}, Thompson {regret_curve(arms_ts)[-1]:.0f}, epsilon-greedy {regret_curve(arms_eg)[-1]:.0f}")
print(f"epsilon-greedy share shown pink in the last 500 visitors: {(arms_eg[:, -500:] == 0).mean():.3f}")

# check: with alpha = 1/n the update Q <- Q + alpha (r - Q) is the running mean
rewards = np.array([1, 0, 0, 1, 1, 0, 1])
Q_run = 0.0
for n_seen, r in enumerate(rewards, start=1):
    Q_run += (1 / n_seen) * (r - Q_run)
print("running update with alpha = 1/n:", round(Q_run, 4), " plain mean:", round(rewards.mean(), 4))
```

**Reading.**

- Epsilon-greedy loses 34 purchases.
- With probability 0.1 it picks a page at random, and half of those picks are black.
- So black gets 5% of the traffic forever, and the regret keeps growing slowly.
- The last line confirms that $\alpha = 1/n$ gives the plain mean, 0.5714 both ways.

## 7. Importance sampling: judging a new rule with old data

- The A/B test logged visitors under a 50/50 rule.
- We want to know how a different rule, such as "always show pink", would have done, without running it.

- A policy is a rule that gives a probability to each action. Here an action is a page.
- The old policy showed pink with probability 0.5 and black with probability 0.5.
- A new policy might show pink with probability 1.
- We evaluate a fixed rule whose action probabilities we know.
- We are not replaying the whole learning path of a Thompson sampler. That sampler would have updated on different visitors and chosen different pages.

- Reweight the logged visitors. Each one gets the weight

$$ w = \frac{\text{chance the new rule shows this page}}{\text{chance the old rule showed this page}}. $$

- A made-up table of four logged visitors, for "always pink" under the old 50/50 rule. Not the SimCity data.

| page shown | reward $r$ | old probability | new probability | weight $w$ | $w \times r$ |
|---|---|---|---|---|---|
| pink | 1 | 0.5 | 1 | 2 | 2 |
| pink | 0 | 0.5 | 1 | 2 | 0 |
| black | 1 | 0.5 | 0 | 0 | 0 |
| black | 0 | 0.5 | 0 | 0 | 0 |

- The estimate of the new rule's purchase rate is the average of $w \times r$ over all logged visitors.
- Here $r$ is 1 for a buy and 0 otherwise.
- The black rows have weight 0, but they still count in the denominator, the number of visitors, because they were logged.
- In the table the average is $(2 + 0 + 0 + 0)/4 = 0.5$.

- Support condition: if the new rule might choose an action, the old rule must sometimes have chosen it.
- Otherwise the weight divides by zero, and the log says nothing about that action.

- We know the truth in this simulation: always pink earns 0.40.


```python
# Simulated log of an A/B test that showed pink or black with probability 0.5 each.
# True purchase rates (known only because this is a simulation): pink 0.40, black 0.30.
true_rates = np.array([0.40, 0.30])          # index 0 = pink, 1 = black
rng_log = np.random.default_rng(7)
n_log = 2000
log_page = rng_log.integers(0, 2, n_log)      # 0 = pink shown, 1 = black shown
log_buy = (rng_log.random(n_log) < true_rates[log_page]).astype(float)
print("visitors logged:", n_log, " shown pink:", int((log_page == 0).sum()), " shown black:", int((log_page == 1).sum()))
print("observed purchase rate, pink visitors :", round(log_buy[log_page == 0].mean(), 3))
print("observed purchase rate, black visitors:", round(log_buy[log_page == 1].mean(), 3))
```

```python
w_pink = np.where(log_page == 0, 1 / 0.5, 0.0)            # new rule: always pink, old rule: 50/50
est_always_pink = (w_pink * log_buy).mean()
print(f"importance-sampling estimate of 'always pink': {est_always_pink:.3f}   (truth: 0.400)")
print(f"naive average over all visitors (what a 50/50 mix earns): {log_buy.mean():.3f}")
print(f"average over pink visitors only: {log_buy[log_page == 0].mean():.3f}")
```

**Reading.**

- The estimate for "always pink" is 0.383, close to the truth 0.40.
- The naive average over all visitors, 0.344, answers a different question: what the 50/50 mix earns.
- With this rule only pink visitors count, each with weight 2, to make up for the black visitors we dropped.
- So the estimate is close to the average over pink visitors, 0.387.
- It equals that average only if exactly half the visitors saw pink. Here 990 of the 2,000 did.

```python
def repeat_is(pink_prob_old, pink_prob_new, n=2000, reps=1000, seed=71):
    """Simulated: reps logs of n visitors under the old rule; importance-sampling estimate of the new rule."""
    rng = np.random.default_rng(seed)
    page = (rng.random((reps, n)) >= pink_prob_old).astype(int)       # 0 = pink, 1 = black
    buy = (rng.random((reps, n)) < true_rates[page]).astype(float)
    new_prob = np.where(page == 0, pink_prob_new, 1 - pink_prob_new)
    old_prob = np.where(page == 0, pink_prob_old, 1 - pink_prob_old)
    return (new_prob / old_prob * buy).mean(axis=1)                    # one estimate per log

est_pink = repeat_is(0.5, 1.0)
est_80 = repeat_is(0.5, 0.8)
print(f"always pink, old rule 50/50 : mean {est_pink.mean():.3f}, sd {est_pink.std():.3f}   (truth 0.400)")
print(f"80% pink,    old rule 50/50 : mean {est_80.mean():.3f}, sd {est_80.std():.3f}   (truth 0.380)")
```

**Reading.**

- Over 1,000 repeats, the estimates average 0.401 for "always pink". The truth is 0.400.
- They average 0.381 for the 80% rule. The truth is 0.380.
- The spread of one estimate is about 0.018 and 0.014. Matching the truth on average is called unbiased.

**Predict.** Now suppose the old rule showed pink only 5% of the time. The weights for "always pink" become $1/0.05 = 20$.

- Will the estimate be right on average?
- Will its spread get smaller, stay the same, or get larger?


```python
est_rare = repeat_is(0.05, 1.0, seed=72)
print(f"always pink, old rule shows pink 5% : mean {est_rare.mean():.3f}, sd {est_rare.std():.3f}   (truth 0.400)")
print(f"spread is {est_rare.std() / est_pink.std():.1f} times larger than with the 50/50 log")
print(f"smallest and largest estimate: {est_rare.min():.3f} and {est_rare.max():.3f}")

fig, ax = plt.subplots()
ax.hist(est_pink, bins=40, range=(0.1, 0.8), alpha=0.6, color="C0", label="old rule: pink 50%")
ax.hist(est_rare, bins=40, range=(0.1, 0.8), alpha=0.6, color="C3", label="old rule: pink 5%")
ax.axvline(0.40, color="black", linestyle="--")
ax.set_xlabel("estimate of the purchase rate of 'always pink'"); ax.set_ylabel("number of repeats"); ax.legend()
plt.show()
```

**Reading.**

- The estimate is still right on average, 0.399.
- One estimate is far less reliable: the spread is 0.063, about 3.5 times larger.
- Only about 100 of the 2,000 visitors saw pink, and each counts 20 times.

- You can only learn about a rule from data in which something sometimes did what the rule does.

- In reinforcement learning this is off-policy learning: learning about one policy from the logs of another.
- It returns at home with Benter's $c$ and $q$.

## 8. Regression to the mean, and why randomize

- A group improves after a treatment. Was it the treatment?
- The pilot and clinic examples below are simulated. Washington's plots are a story, with no data.

- In *Thinking, Fast and Slow* (2011), Daniel Kahneman describes flight instructors.
- After they criticized a bad landing, the next landing was usually better. After they praised a good one, the next was usually worse.
- They concluded that criticism works and praise does not.
- We simulate pilots in which praise and criticism have no effect at all.
- Each landing score is the pilot's fixed skill plus independent luck. Higher is better.

**Predict.**

- Take the worst 10% of pilots on landing 1, and look at their landing 2.
- With no feedback effect at all, does their score go up, stay the same, or go down? And the best 10%?


```python
rng_p = np.random.default_rng(1973)
n_pilots = 2000
skill = rng_p.normal(0, 1, n_pilots)
land1 = skill + rng_p.normal(0, 1, n_pilots)       # skill plus luck
land2 = skill + rng_p.normal(0, 1, n_pilots)       # new luck; no effect of praise or criticism is built in
worst = land1 <= np.percentile(land1, 10)
best = land1 >= np.percentile(land1, 90)
for name, grp in [("worst 10% on landing 1", worst), ("best 10% on landing 1", best), ("all pilots", np.ones(n_pilots, bool))]:
    print(f"{name:24s}: landing 1 {land1[grp].mean():+.2f}, landing 2 {land2[grp].mean():+.2f}, change {land2[grp].mean() - land1[grp].mean():+.2f}")
```

**Reading.**

- The worst 10% score -2.48 on landing 1 and -1.33 on landing 2. They improved by +1.15, with no feedback at all.
- The best 10% went from +2.38 to +1.13, a change of -1.26.
- All pilots together do not change, -0.03.
- We selected the extreme pilots on a noisy measurement.
- A very bad landing 1 is partly bad skill and partly bad luck. The skill stays. The luck does not repeat.
- Criticism looked effective only because the instructors picked the worst landings.

- A clinic picks the 20% lowest scorers on a noisy pre-test, higher is better, and gives them a treatment that does nothing.
- We also randomize that same selected group into treatment and control.


```python
rng_r = np.random.default_rng(1981)
n_pat = 4000
level = rng_r.normal(50, 10, n_pat)                # stable true level of each person
pre = level + rng_r.normal(0, 8, n_pat)            # noisy pre-test
post = level + rng_r.normal(0, 8, n_pat)           # noisy retest; the treatment adds exactly 0
low = pre <= np.percentile(pre, 20)                # the selected group: lowest 20% on the pre-test
change = post - pre
print(f"selected group: n = {low.sum()}, pre-test mean {pre[low].mean():.1f}, retest mean {post[low].mean():.1f}, before/after improvement {change[low].mean():+.2f}")

idx = rng_r.permutation(np.flatnonzero(low))       # randomize the SAME selected group
treat, ctrl = idx[: len(idx) // 2], idx[len(idx) // 2:]
est = change[treat].mean() - change[ctrl].mean()
se_est = np.sqrt(change[treat].var(ddof=1) / len(treat) + change[ctrl].var(ddof=1) / len(ctrl))
print(f"randomized halves: treated improved {change[treat].mean():+.2f}, control improved {change[ctrl].mean():+.2f}")
print(f"estimated effect (treated change minus control change) = {est:+.2f}, standard error {se_est:.2f}   (true effect: 0)")

print(f"control drawn from everyone else instead: change {change[~low].mean():+.2f}, so the naive effect estimate is {change[low].mean() - change[~low].mean():+.2f}")
no_sel = rng_r.random(n_pat) < 0.2                 # no selection: a random 20%
print(f"no selection (random 20% of people): change {change[no_sel].mean():+.2f}")
low_level = level <= np.percentile(level, 20)      # selection on a noise-free measurement: level itself, measured twice
print(f"no noise (select on the true level, measured without error): change {np.mean(level[low_level] - level[low_level]):+.2f}")
```

**Reading.**

- The selected group improves by +7.69, from 31.9 to 39.6, although the treatment does nothing.
- After randomizing that same group, both halves improve by about the same amount, +8.04 and +7.34.
- The estimated effect is +0.70, standard error 0.74, consistent with a true effect of zero.
- A control group drawn from everyone else does not fix this.
- Those people were not selected for a low score, so they do not improve, -2.09.
- The naive estimate is then +9.78, entirely spurious.
- The last two lines: no selection, or no noise, gives no improvement, -0.23 and +0.00.

- Two caveats, after K. Hahn, "Regression to the mean is a rhetorical device", *Methodology Matters*: https://methodologymatters.substack.com/p/regression-to-the-mean-is-a-rhetorical
- Regression to the mean is not a law of nature. It comes from selecting on a noisy measurement.
- When a group improves without treatment, name the reason.
- "They might have improved anyway" is a specific claim: the natural course, with luck that did not repeat.
- "It's just regression to the mean" is not an answer by itself.

**Check yourself.** Why does a control group drawn from the same selected units remove the bias, and why would a control group drawn from everyone not remove it?

<details><summary>Answer</summary>

- Both halves were selected for a low pre-test, so both have the same luck to lose.
- The improvement due to luck is in both changes, and it cancels in the difference.
- A control drawn from everyone was not selected, so its change has no such improvement.
- The difference then keeps the luck, as if it were a treatment effect.

</details>


### Washington's plots

## 9. Russian election observers

- In the Moscow elections to the State Duma on 4 December 2011, Enikolopov and coauthors (2013, *PNAS*) randomly sent observers.
- Independent observers went to 156 of the 3,164 polling stations in Moscow.
- They compared the reported vote share of the ruling party, United Russia, at stations with and without observers.
- Chance assignment balances other differences between stations on average.
- United Russia's share was about 11 percentage points lower where observers were present.

- The data below are simulated. We assume a control mean share of 0.46 and a spread across stations of 0.12.
- Those two numbers are made up for the simulation. They are not the paper's.


```python
rng_ru = np.random.default_rng(2011)
n_stations, n_obs = 3164, 156
has_observer = np.zeros(n_stations, dtype=bool)
has_observer[rng_ru.choice(n_stations, n_obs, replace=False)] = True        # assigned at random
share = np.clip(0.46 - 0.11 * has_observer + rng_ru.normal(0, 0.12, n_stations), 0, 1)   # simulated vote share

diff_obs = share[has_observer].mean() - share[~has_observer].mean()
se_ru = np.sqrt(share[has_observer].var(ddof=1) / n_obs + share[~has_observer].var(ddof=1) / (n_stations - n_obs))
print(f"simulated mean share, with observers {share[has_observer].mean():.3f}, without {share[~has_observer].mean():.3f}")
print(f"difference = {diff_obs:.3f}, standard error = {se_ru:.4f}, 95% interval ({diff_obs - 1.96 * se_ru:.3f}, {diff_obs + 1.96 * se_ru:.3f})")
```

**Reading.**

- In the simulation, stations with observers have a mean share of 0.340, and the others 0.461.
- The difference is -0.121, with a 95% interval of (-0.139, -0.103).
- The made-up true gap is -0.11, and the simulated gap is -0.121.
- The [Rent Control Lab](https://vsokolov.org/labs/homeless-lab.html) also has a permutation test.

**Your turn: a permutation test.**

- The null says the observer made no difference, so the labels "observer" and "no observer" could be shuffled freely.
- Shuffle the labels many times, compute the difference of means each time, and see where the real difference falls.
- Change `n_perm` and look at the spread.


```python
n_perm = 2000   # change this
rng_pm = np.random.default_rng(9)
diffs = np.empty(n_perm)
for i in range(n_perm):
    shuffled = rng_pm.permutation(has_observer)                 # same number of observer stations, random ones
    diffs[i] = share[shuffled].mean() - share[~shuffled].mean()
n_extreme = int((np.abs(diffs) >= abs(diff_obs)).sum())     # np.abs: two-sided, a gap of either sign counts
p_perm = (n_extreme + 1) / (n_perm + 1)                       # never report exactly 0 from a finite number of shuffles
print(f"shuffled differences: sd {diffs.std():.4f}, range ({diffs.min():.3f}, {diffs.max():.3f})")
print(f"real difference {diff_obs:.3f}; shuffles at least as extreme: {n_extreme} of {n_perm}")
print(f"permutation p-value (count + 1)/(n_perm + 1) = {p_perm:.4f}")
```

**Reading.**

- The shuffled differences have standard deviation 0.0101, and they stay within about 0.04 of zero.
- The real difference, -0.121, is far outside. It is in none of the 2,000 shuffles.
- So the p-value estimate is $(0 + 1)/(2000 + 1) = 0.0005$. These numbers are for `n_perm = 2000`, as given.
- We use `np.abs`, so the test is two-sided: a shuffled gap of either sign counts as extreme.


## 10. Gujarat pollution audits (simulated)

- Duflo and coauthors (2013) ran a randomized experiment in Gujarat, India.
- Plants paid their own environmental auditors.
- Auditors tended to report pollution just under the legal limit, even when the true value was higher.
- In the treatment group, auditors were assigned at random, paid from a central pool, and checked by random backchecks.
- A backcheck is an independent repeat measurement of the same plant by a technical agency, used to catch false reports.

- This is the same story as the lemons problem in Lab 8.
- When the seller pays the inspector, the inspection says what the seller wants.

- The data below are simulated, not from the paper.
- True emissions of 400 plants, median about 130, against a legal limit of 150.
- In the control group, a report is moved to just under the limit, with probability `p_fudge`, when the truth is above the limit.
- In the treatment group, reports are honest.
- We look for bunching: a pile-up of reports just below 150.


```python
rng_g = np.random.default_rng(94)
limit = 150
truth = rng_g.lognormal(mean=np.log(130), sigma=0.35, size=400)
p_fudge = 0.8   # change this
fudge = (truth > limit) & (rng_g.random(400) < p_fudge)
report_control = np.where(fudge, rng_g.uniform(0.85 * limit, limit - 0.5, 400), truth)    # auditor paid by the plant
report_treat = truth.copy()                                                              # honest reports

band = lambda x: ((x >= 0.75 * limit) & (x < limit)).mean()
print(f"share of reports between 75% and 100% of the limit: control {band(report_control):.3f}, treatment {band(report_treat):.3f}")
print(f"share of reports above the limit: control {(report_control >= limit).mean():.3f}, treatment {(report_treat >= limit).mean():.3f}")
bins = np.arange(40, 360, 10)
plt.hist(report_treat, bins=bins, alpha=0.6, color="C0", label="treatment (simulated)")
plt.hist(report_control, bins=bins, alpha=0.6, color="C3", label="control (simulated)")
plt.axvline(limit, color="black", linestyle="--")
plt.xlabel("reported pollution (mg per cubic meter)"); plt.ylabel("number of plants"); plt.legend()
plt.show()
```

**Reading.**

- 0.645 of control reports lie between 75% and 100% of the limit, against 0.365 in the treatment group.
- Reports above the limit are 0.080 for control and 0.360 for treatment, although the true emissions are the same in both.
