```python
import copy
from hashlib import sha256
from pathlib import Path
from urllib.request import urlretrieve

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch
from sklearn.ensemble import GradientBoostingRegressor, RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor
from torch import nn
from torch.utils.data import DataLoader, TensorDataset

DATA = Path("data")
DATA.mkdir(exist_ok=True)
BASE = "https://vsokolov.org/courses/files/600/data"
for name in ("diamonds.csv", "circle.csv", "SaratogaHouses.csv"):
    if not (DATA / name).exists():
        urlretrieve(f"{BASE}/{name}", DATA / name)

PREFIX = "aii600-trees-nn-sgd"


def check(name, value, expected, ndigits=6):
    if value is None:
        raise AssertionError(f"not yet: {name} is still None")
    try:
        x = float(value)
    except (TypeError, ValueError):
        raise AssertionError(f"not yet: {name} must be a plain number, not {type(value).__name__}")
    s = f"{round(x, ndigits):.{ndigits}f}"
    got = sha256(f"{PREFIX}|{name}|{s}".encode()).hexdigest()[:16]
    assert got == expected, f"not yet: {name} is not right (yours: {x:.6g})"
    print("ok ", name)


def check_array(name, value, expected, shape, ndigits=6):
    if value is None:
        raise AssertionError(f"not yet: {name} is still None")
    arr = np.asarray(value, dtype=float)
    if arr.shape != tuple(shape):
        raise AssertionError(f"not yet: {name} has shape {arr.shape}, the task asks for {tuple(shape)}")
    key = (np.round(arr, ndigits) + 0.0).tobytes()
    got = sha256(f"{PREFIX}|{name}|".encode() + key).hexdigest()[:16]
    assert got == expected, f"not yet: {name} has the right shape but some values are wrong"
    print("ok ", name)


def check_near(name, value, target, tol, relative=False):
    # for library results that can differ a little between versions
    if value is None:
        raise AssertionError(f"not yet: {name} is still None")
    gap = abs(float(value) - target) / (abs(target) if relative else 1.0)
    assert gap < tol, f"not yet: {name} is not within the tolerance of the right value (yours: {float(value):.6g})"
    print("ok ", name)


def sigmoid(z):
    return 1 / (1 + np.exp(-z))


def rmse(a, b):
    return float(np.sqrt(np.mean((np.asarray(a) - np.asarray(b)) ** 2)))
```

### Exercise 2: information gain, twenty questions with chess pieces

This exercise introduces entropy, a second way (besides SSE) to score a split.

- Entropy: uncertainty about the class in a node, in bits. $H = -\sum_k p_k \log_2 p_k$, where $p_k$ is the share of class $k$.
- One bit: the uncertainty of one fair yes-or-no question. Two equally likely classes: $H = -2 \cdot \frac12 \log_2 \frac12 = 1$.
- $m$ equally likely classes: $H = \log_2 m$. One class only (a pure node): $H = 0$.
- Information gain of a question: entropy before minus the size-weighted entropy after. A tree asks the question with the largest gain.

**Task.** Six chess pieces (king, queen, rook, bishop, knight, pawn), each equally likely. Score two yes-or-no questions by information gain.

**Inputs.** None.

**Formulas.** Information gain of a split into groups L and R: $H(\text{node}) - \frac{n_L}{n}H(L) - \frac{n_R}{n}H(R)$, where $n_L, n_R$ are the group sizes and $n = 6$.

- Question A splits the pieces 3 and 3.
- Question B sets the knight apart: 1 and 5.

**Define.** `H_root` (entropy of all six pieces), `gain_33` (gain of A), `gain_15` (gain of B). Bits, do not round.

### Exercise 5: why averaging trees has a floor

**Task.** Compute the mean squared error of an average of $B$ trees.

**Inputs.** `sigma2_tree` (mean squared error of one tree), `rho_bag` (correlation of two bagged trees' errors), `rho_forest` (the same for a random forest), `B_list`.

**Formula.** Average of $B$ trees: $\text{MSE}_B = \rho\,\sigma^2 + \frac{1-\rho}{B}\,\sigma^2$, where $\sigma^2$ is one tree's MSE and $\rho$ the correlation. As $B$ grows, $\text{MSE}_B \to \rho\sigma^2$.

- `mse_forest`: $\text{MSE}_B$ for the forest at each $B$ in `B_list`, shape (3,).
- `floor_bag`, `floor_forest`: the limits $\rho\sigma^2$.


```python
sigma2_tree = 4.0
rho_bag, rho_forest = 0.6, 0.25
B_list = np.array([1, 10, 500])
```

### Exercise 7: needles in a haystack, p much larger than n

**Task.** 100 patients, 1,000 genes, 10 needles (true effect 1) and 990 hay genes (effect 0). Compare one-gene-at-a-time testing with the lasso.

**Inputs.** `X_g` (100 × 1000), `y_g` (100,), `needles` (indices of the 10 needles), `hay` (True for hay genes, shape (1000,)), `marginal_pvalues(X, y)` (p-value of each gene from its own simple regression), `Lasso` (scikit-learn).

**Steps.**
1. Screening: `pv = marginal_pvalues(X_g, y_g)`. Count hay genes with $p < 0.05$ (false alarms; expect about 5% of 990).
2. Lasso: `Lasso(alpha=lam_univ, max_iter=100_000, tol=1e-10).fit(X_g, y_g)`, with `lam_univ` from Exercise 6. A gene is selected when its coefficient in `.coef_` is not 0.

**Define.** `hay_sig` (hay genes with $p < 0.05$), `needles_found` (needles selected by the lasso), `hay_selected` (hay genes selected by the lasso). Integers.

**Tolerance.** Each count within 1 (scikit-learn versions differ a little).


```python
from scipy import stats
from sklearn.linear_model import Lasso

rng_g = np.random.default_rng(4)
X_g = rng_g.standard_normal((100, 1000))
needles = np.arange(10) * 100
beta_g = np.zeros(1000)
beta_g[needles] = 1.0
y_g = X_g @ beta_g + rng_g.standard_normal(100)
hay = np.ones(1000, bool)
hay[needles] = False

def marginal_pvalues(X, y):
    # two-sided p-value of the slope in y = a + b x_j, for each column j
    n = len(y)
    Xc, yc = X - X.mean(axis=0), y - y.mean()
    b = Xc.T @ yc / (Xc**2).sum(axis=0)
    resid = yc[:, None] - Xc * b
    se = np.sqrt((resid**2).sum(axis=0) / (n - 2) / (Xc**2).sum(axis=0))
    return 2 * stats.t.sf(np.abs(b / se), n - 2)

print(X_g.shape, "rank of X'X:", np.linalg.matrix_rank(X_g), "of 1000")
```

```python
# Exercise 7: define hay_sig, needles_found, hay_selected
hay_sig = None
needles_found = None
hay_selected = None
```

```python
check_near("hay_sig", hay_sig, 46, 1.01)
check_near("needles_found", needles_found, 10, 1.01)
check_near("hay_selected", hay_selected, 10, 1.01)
```

```python
# Plot: one gene at a time (left) against the lasso (right). Red: the 10 needles.
pv_plot = marginal_pvalues(X_g, y_g)
coef_plot = Lasso(alpha=lam_univ, max_iter=100_000, tol=1e-10).fit(X_g, y_g).coef_
fig, axes = plt.subplots(1, 2, figsize=(10, 3.5))
axes[0].scatter(np.where(hay)[0], -np.log10(pv_plot[hay]), s=6, color="grey", label="hay")
axes[0].scatter(needles, -np.log10(pv_plot[needles]), s=25, color="C3", label="needles")
axes[0].axhline(-np.log10(0.05), color="k", ls="--", label="p = 0.05")
axes[0].set_xlabel("gene"); axes[0].set_ylabel("-log10 p-value"); axes[0].set_title(f"{hay_sig} hay genes cross p = 0.05"); axes[0].legend(frameon=False, fontsize=8)
axes[1].vlines(np.where(hay)[0], 0, coef_plot[hay], color="grey")
axes[1].vlines(needles, 0, coef_plot[needles], color="C3", lw=2)
axes[1].axhline(0, color="k", lw=0.5)
axes[1].set_xlabel("gene"); axes[1].set_ylabel("lasso coefficient"); axes[1].set_title(f"lasso: {needles_found} needles, {hay_selected} hay")
plt.tight_layout(); plt.show()
```

### Exercise 12: the step-size ceiling

**Task.** For the line $\theta_1 + \theta_2 x$ with loss $L(\theta) = \frac{1}{2n}\sum_i (y_i - \theta_1 - \theta_2 x_i)^2$, find the largest step size for which gradient descent converges, with $x$ centred at 0 and with $x$ shifted to mean 3.

### Exercise 13: constant steps hover, shrinking steps converge

**Task.** Estimate a mean $\theta$ by SGD with batch size 1. The loss of one example is $\frac12(\theta - y_i)^2$, so its gradient is $\theta - y_i$. Run 5,000 steps from $\theta_0 = 0$, using example `idx_sgd[t]` at step $t$, three ways.

**Inputs.** `y_sgd` (200 values), `idx_sgd` (5,000 example indices), `eta0`.

**Formulas.** Step $t = 0, 1, \ldots, 4999$: $\theta_{t+1} = \theta_t - \eta_t\,(\theta_t - y_{\text{idx}[t]})$.
